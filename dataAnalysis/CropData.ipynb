```python
from dataclasses import dataclass, field
 
import numpy as np
import pandas as pd
```

```python
YIELD_REQUIRED = ["crop", "year", "season", "state", "area", "production","fertilizer", "pesticide", "yield"]

SOIL_REQUIRED = ["state", "N", "P", "K", "pH"]

WEATHER_REQUIRED = ["state", "year", "avg_temp_c", "total_rainfall_mm","avg_humidity_percent"]

KEY = ["crop", "state", "season", "year"]

NUMERIC_YIELD = ["area", "production", "fertilizer", "pesticide", "yield"]
```

```python
# Columns that uniquely identify one yield record
KEY: list[str] = ["crop", "state", "season", "year"]

def norm_text(s: pd.Series) -> pd.Series:
    """Trim, collapse repeated spaces and title-case, so keys match across tables."""
    return (
        s.astype("string")
         .str.strip()                            # "Kerala "      -> "Kerala"
         .str.replace(r"\s+", " ", regex=True)   # "Tamil   Nadu" -> "Tamil Nadu"
         .str.title()                            # "KERALA"       -> "Kerala"
    )   
```

```python
y: pd.DataFrame = df_yield.copy()   # never mutate the raw data

# Fix whitespace and casing in categorical columns
for c in ["crop", "season", "state"]:
    y[c] = norm_text(y[c])

# Force numeric types; bad strings become NaN instead of raising
for c in ["year", "area", "production", "fertilizer", "pesticide", "yield"]:
    y[c] = pd.to_numeric(y[c], errors="coerce")
```

```python
y
```

```python
# Rows missing a key, or the two columns yield depends on, are unusable
y = y.dropna(subset=KEY + ["area", "production"])

# Exact duplicate rows add nothing
y = y.drop_duplicates()

# Zero area or production means a failed record, not a real yield
y = y[(y["area"] > 0) & (y["production"] > 0)]

# year became float after coercion; restore integer type
y["year"] = y["year"].astype(int)
```

```python
y = (
    y.sort_values("area", ascending=False)   # largest area first
     .drop_duplicates(KEY, keep="first")     # keep that row for each key
     .sort_values(KEY)
     .reset_index(drop=True)
)

assert not y.duplicated(KEY).any()           # key is now unique
```

```python
y
```

```python
# Recompute rather than trust the supplied column
y["yield"] = y["production"] / y["area"]

# Fertilizer and pesticide are farm totals and mostly encode farm size,
# so per-hectare values are the comparable ones
y["fertilizer_per_ha"] = y["fertilizer"] / y["area"]
y["pesticide_per_ha"] = y["pesticide"] / y["area"]
```

```python
soil: pd.DataFrame = df_soil.copy()
weather: pd.DataFrame = df_weather.copy()

# Same normalisation as the yield table, otherwise joins silently miss
soil["state"] = norm_text(soil["state"])
weather["state"] = norm_text(weather["state"])

# Join keys must be unique on the lookup side
assert not soil["state"].duplicated().any(), "duplicate states in soil"
assert not weather.duplicated(["state", "year"]).any(), "duplicate state-year in weather"

# Every state in the yield table must exist in both lookups
miss_soil = set(y["state"]) - set(soil["state"])
miss_wx = set(y["state"]) - set(weather["state"])
assert not miss_soil, f"states missing from soil: {sorted(miss_soil)}"
assert not miss_wx, f"states missing from weather: {sorted(miss_wx)}"
```

```python
feat: list[str] = ["N", "P", "K", "pH", "avg_temp_c","total_rainfall_mm", "avg_humidity_percent"]

df: pd.DataFrame = (
    # weather is one row per state-year; validate stops row multiplication
    y.merge(weather, on=["state", "year"], how="left", validate="m:1")
    # soil is one row per state
     .merge(soil, on="state", how="left", validate="m:1")
)

# Left joins leave NaN wherever a key had no match
assert df[feat].notna().all().all(), "some rows have no soil or weather match"
```

```python
# Crops with too few records give unstable medians, so drop them first
MIN_CROP_ROWS: int = 100
# Argmax over 1-2 crops is meaningless; require a real choice set
MIN_CROPS_PER_GROUP: int = 3
# One recommendation is made per state, season and year
GROUP: list[str] = ["state", "season", "year"]

crop_counts: pd.Series = df["crop"].value_counts()
keep_crops = crop_counts[crop_counts >= MIN_CROP_ROWS].index

d: pd.DataFrame = df[df["crop"].isin(keep_crops)].copy()
```

```python
# Max yield is ~21,000 against a median near 1, so unclipped values would
# decide the argmax by themselves. Clip within each crop at the 1st/99th percentile.
by_crop = d.groupby("crop")["yield"]
lo = by_crop.transform(lambda s: s.quantile(0.01))
hi = by_crop.transform(lambda s: s.quantile(0.99))

d["yield_clipped"] = d["yield"].clip(lower=lo, upper=hi)
```

```python
# Raw yields aren't comparable across crops (different units).
# Dividing by the crop's own median gives "how good was this vs. typical for this crop".
crop_median = d.groupby("crop")["yield_clipped"].transform("median")
d["rel_yield"] = d["yield_clipped"] / crop_median

assert (crop_median > 0).all(), "a crop has a zero median yield"
```

```python
# How many crops were actually grown in each state-season-year
d["n_candidates"] = d.groupby(GROUP)["crop"].transform("size")
d = d[d["n_candidates"] >= MIN_CROPS_PER_GROUP]

# Row index of the highest relative yield in each group
best_idx = d.groupby(GROUP)["rel_yield"].idxmax()

best: pd.DataFrame = (
    d.loc[best_idx, GROUP + ["crop"]]
     .rename(columns={"crop": "recommended_crop"})
)
```

```python
best
```

```python
feat_cols: list[str] = ["N", "P", "K", "pH", "avg_temp_c",
                        "total_rainfall_mm", "avg_humidity_percent"]

# Soil and weather are identical for every crop in a group, so take the first value
context: pd.DataFrame = d.groupby(GROUP, as_index=False)[feat_cols].first()

data: pd.DataFrame = context.merge(best, on=GROUP, validate="1:1")

assert len(data) == len(best), "group count changed in the join"
```

```python
# Rebuild from the unmerged labels, since `data["recommended_crop"]` was overwritten
data = context.merge(best, on=GROUP, validate="1:1")

MIN_CLASS_ROWS: int = 30   # 100 was too strict for one row per state-season-year

counts: pd.Series = data["recommended_crop"].value_counts()
rare = counts[counts < MIN_CLASS_ROWS].index
data["recommended_crop"] = data["recommended_crop"].where(
    ~data["recommended_crop"].isin(rare), "Other"
)

# Guards: top-3 only means something with enough classes, and no class should dominate
shares: pd.Series = data["recommended_crop"].value_counts(normalize=True)
assert shares.shape[0] >= 8, "fewer than 8 classes; a top-3 shortlist is trivial"
assert shares.iloc[0] < 0.5, "one class still dominates the labels"

shares.round(3)   # displays the new class balance
```

```python
# Rebuild the labels first by rerunning Step 14b; then remove the catch-all bucket.
# "Other" is not a crop, and it inflates the baseline (always predicting it scores 25%).
data = data[data["recommended_crop"] != "Other"].reset_index(drop=True)
```

```python
from sklearn.preprocessing import LabelEncoder

# Integer label for the model; keep `le` to decode predictions later
le = LabelEncoder()
data["label"] = le.fit_transform(data["recommended_crop"].astype(str))

# One-hot season (a handful of categories, no order)
data = pd.get_dummies(data, columns=["season"], prefix="season", dtype=int)
```

```python
season_cols = [c for c in data.columns if c.startswith("season_")]
X_cols: list[str] = feat_cols + season_cols

X: pd.DataFrame = data[X_cols]
y_target: pd.Series = data["label"]     # named y_target so the cleaned `y` table isn't overwritten
groups: pd.Series = data["state"]       # for GroupKFold; NOT a feature
years: pd.Series = data["year"]         # for the time split; NOT a feature

assert X.notna().all().all(), "missing values in features"
assert y_target.nunique() >= 2, "only one class survived filtering"

# These build or leak the label and must never be inputs
leaky = {"area", "production", "yield", "fertilizer", "pesticide",
         "fertilizer_per_ha", "pesticide_per_ha", "state", "year", "label"}
assert not leaky & set(X.columns), "leaky column in X"
```

```python
import numpy as np
import joblib
from lightgbm import LGBMClassifier
from sklearn.base import clone
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import (ExtraTreesClassifier,
                              HistGradientBoostingClassifier,
                              RandomForestClassifier)
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, top_k_accuracy_score
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

SEED: int = 42
N_CLASSES: int = y_target.nunique()

# Labels must be 0..N-1 so probability columns line up with class ids
assert set(y_target.unique()) == set(range(N_CLASSES)), "labels are not contiguous"
assert N_CLASSES >= 3, "top-3 accuracy needs at least 3 classes"
assert groups.nunique() >= 5, "need at least 5 states for 5-fold GroupKFold"

# Each fold holds out whole states, so scores reflect new regions, not memorised ones
cv = GroupKFold(n_splits=5)
```

```python
# Distance- and linear-based models need scaled inputs; tree models do not.
# class_weight="balanced" offsets the uneven crop classes.
# HistGradientBoostingClassifier's class_weight needs scikit-learn >= 1.2.
models: dict[str, object] = {
    "baseline_prior": DummyClassifier(strategy="prior"),   # must be beaten

    "logistic_regression": make_pipeline(
        StandardScaler(),
        LogisticRegression(max_iter=2000, class_weight="balanced"),
    ),
    "knn": make_pipeline(
        StandardScaler(),
        KNeighborsClassifier(n_neighbors=25),
    ),
    "random_forest": RandomForestClassifier(
        n_estimators=400, min_samples_leaf=3, class_weight="balanced",
        n_jobs=-1, random_state=SEED,
    ),
    "extra_trees": ExtraTreesClassifier(
        n_estimators=400, min_samples_leaf=3, class_weight="balanced",
        n_jobs=-1, random_state=SEED,
    ),
    "hist_gradient_boosting": HistGradientBoostingClassifier(
        class_weight="balanced", random_state=SEED,
    ),
    "lightgbm": LGBMClassifier(
        n_estimators=400, learning_rate=0.05, class_weight="balanced",
        random_state=SEED, n_jobs=-1, verbose=-1,          # verbose=-1 silences output
    ),
}
```

```python
def evaluate(model: object, X: pd.DataFrame, y: pd.Series,
             groups: pd.Series) -> dict[str, float]:
    """Out-of-fold scores with whole states held out in each fold."""
    # Out-of-fold probabilities: every row is scored by a model that never saw its state
    proba: np.ndarray = cross_val_predict(
        model, X, y, groups=groups, cv=cv, method="predict_proba"
    )
    labels = np.arange(N_CLASSES)
    return {
        "top1": top_k_accuracy_score(y, proba, k=1, labels=labels),
        "top3": top_k_accuracy_score(y, proba, k=3, labels=labels),   # shortlist quality
        "macro_f1": f1_score(y, proba.argmax(axis=1), average="macro"),  # treats classes equally
    }


# This is the slow cell: 7 models x 5 folds
results: pd.DataFrame = (
    pd.DataFrame({name: evaluate(m, X, y_target, groups) for name, m in models.items()})
      .T
      .sort_values(["top3", "macro_f1"], ascending=False)   # rank by top-3, then macro F1
)

results.round(3)   # displays the comparison table
```

```python
# Rank excluding the baseline, which exists only as a floor
best_name: str = results.drop(index="baseline_prior").index[0]

# If this fails, the features carry no signal beyond "most common crop"
assert results.loc[best_name, "top3"] > results.loc["baseline_prior", "top3"], \
    "best model does not beat the baseline"

best_name   # displays the winner
```

```python
# A second, different test: train on the past, test on the most recent years
train_mask: pd.Series = years <= 2015
test_mask: pd.Series = years >= 2016
assert train_mask.any() and test_mask.any(), "empty train or test period"

m = clone(models[best_name]).fit(X[train_mask], y_target[train_mask])

# The training period may miss a class; place its probabilities in full-width columns
proba_part: np.ndarray = m.predict_proba(X[test_mask])
proba_full: np.ndarray = np.zeros((proba_part.shape[0], N_CLASSES))
proba_full[:, m.classes_] = proba_part

y_test = y_target[test_mask]
{
    "top1": top_k_accuracy_score(y_test, proba_full, k=1, labels=np.arange(N_CLASSES)),
    "top3": top_k_accuracy_score(y_test, proba_full, k=3, labels=np.arange(N_CLASSES)),
    "macro_f1": f1_score(y_test, proba_full.argmax(axis=1), average="macro"),
}
```

```python
# Refit the winner on all data
final_model = clone(models[best_name]).fit(X, y_target)

# Save everything needed to predict later, in one file
artifact: dict[str, object] = {
    "model": final_model,
    "label_encoder": le,        # decodes class ids back to crop names
    "features": X_cols,         # exact column order the model expects
    "model_name": best_name,
}
joblib.dump(artifact, "crop_recommender.joblib")


def recommend_top_k(features: pd.DataFrame, k: int = 3) -> list[list[str]]:
    """Return the k most likely crops per row, best first."""
    # Selecting by name raises KeyError if a column (e.g. a season dummy) is missing
    proba: np.ndarray = final_model.predict_proba(features[X_cols])
    top: np.ndarray = np.argsort(proba, axis=1)[:, ::-1][:, :k]   # indices, high to low
    return [list(le.inverse_transform(final_model.classes_[row])) for row in top]


recommend_top_k(X.iloc[[0]])   # quick smoke test on one row
```

```python
# `best` still holds the labels from before the rare-class merge
pre_merge: pd.Series = best["recommended_crop"].value_counts()

# After the merge: expect about 3 classes with one near 0.86
post_merge: pd.Series = data["recommended_crop"].value_counts(normalize=True).round(3)

pre_merge.head(15), post_merge
```

```python
def fold_top3(model: object) -> np.ndarray:
    """Top-3 accuracy on each held-out-state fold."""
    scores: list[float] = []
    for tr, te in cv.split(X, y_target, groups):
        m = clone(model).fit(X.iloc[tr], y_target.iloc[tr])
        # A fold's training states may miss a class; fill full-width probabilities
        p: np.ndarray = np.zeros((len(te), N_CLASSES))
        p[:, m.classes_] = m.predict_proba(X.iloc[te])
        scores.append(top_k_accuracy_score(y_target.iloc[te], p, k=3,
                                           labels=np.arange(N_CLASSES)))
    return np.array(scores)


spread: pd.DataFrame = pd.DataFrame({
    n: fold_top3(models[n]) for n in ["baseline_prior", "lightgbm", "hist_gradient_boosting"]
})

# If the std is about as large as the gap between means, the models aren't distinguishable
spread.agg(["mean", "std"]).round(3)
```
